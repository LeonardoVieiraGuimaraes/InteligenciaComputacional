# Raciocínio Fuzzy - 4 Etapas

Este notebook descreve as quatro etapas principais do raciocínio fuzzy.

## 1. Graus de pertinência

Compare os fatos conhecidos com os antecedentes das regras fuzzy para encontrar os graus de pertinência em relação a cada função do antecedente.

## 2. Força de ativação

Combine os graus de pertinência dos antecedentes de cada regra usando operadores AND ou OR para formar uma força de ativação que indica o grau em que a parte antecedente da regra é satisfeita.

## 3. Implicação do consequente

Aplique a força de ativação ao consequente da regra para gerar o grau de ativação da regra.

## 4. Saída

Agregue todos os consequentes ativados para obter uma saída geral do sistema fuzzy.

### Operadores típicos em sistemas fuzzy

- **AND**: usualmente utiliza-se uma T-norma (por exemplo, mínimo) para calcular a força de ativação de uma regra com AND nos antecedentes.
- **OR**: usualmente utiliza-se uma S-norma (por exemplo, máximo) para calcular a força de ativação de uma regra com OR nos antecedentes.
- **Operador de Implicação**: normalmente uma T-norma é usada para calcular o consequente com base na força de disparo.
- **Operador de Agregação**: geralmente uma S-norma é utilizada para agregar os consequentes e gerar a saída geral.
- **Operador de Defuzzificação**: converte a função de pertinência de saída em um valor crisp ou escalar.

## Métodos de Inferência

Até agora, discutimos sistemas baseados em regras fuzzy em que tanto os antecedentes quanto os consequentes são conjuntos fuzzy.

---

Hoje veremos um novo tipo de sistema baseado em regras fuzzy, no qual os antecedentes são conjuntos fuzzy, mas os consequentes são funções das variáveis independentes (por exemplo, funções lineares). Esse tipo de sistema é conhecido como sistema fuzzy do tipo Takagi-Sugeno-Kang (TSK).

## Modelo de TSK (Takagi-Sugeno-Kang)

O modelo TSK é semelhante ao método de Mamdani em vários aspectos: utiliza fuzzificação das entradas, aplicação de operadores fuzzy e uma base de regras condicionais para inferência.

No entanto, a principal diferença está nos consequentes das regras: em vez de serem expressos por conjuntos fuzzy linguísticos, os consequentes do modelo TSK são funções paramétricas (por exemplo, funções lineares) que relacionam as entradas e as saídas do processo. Ou seja, os consequentes são funcionais e não linguísticos.

Essa diferença implica em uma sintaxe distinta para o banco de conhecimento em relação ao modelo Mamdani, o que também afeta as propriedades semânticas do sistema, como será detalhado posteriormente.

## Estrutura das Regras no Modelo TSK

No modelo TSK, as regras têm a seguinte forma:

> **Se X é A<sub>i</sub> e Y é B<sub>i</sub> então z = f<sub>i</sub>(x, y)**

- A<sub>i</sub>, B<sub>i</sub>: conjuntos fuzzy dos antecedentes.
- O consequente é uma função das variáveis de entrada.

### Tipos de modelos TSK

- **Ordem um:** f<sub>i</sub>(x, y) é um polinômio de ordem 1 (linear), por exemplo: z = a·x + b·y + c
- **Ordem zero:** f<sub>i</sub>(x, y) é uma constante, por exemplo: z = c (a = b = 0)

## Exemplo 1: TSK de ordem zero

Considere as regras:

1. Se X é Baixo e Y é Baixo então z = 2  
2. Se X é Alto e Y é Alto então z = 8

**Passos:**
1. Fuzzifique as entradas X e Y para obter os graus de pertinência em "Baixo" e "Alto".
2. Calcule a força de ativação de cada regra (usando, por exemplo, o operador mínimo).
3. O consequente de cada regra é uma constante (2 ou 8).
4. A saída final é a média ponderada das constantes, ponderada pelas forças de ativação.

**Fórmula da saída:**

z = (w₁·z₁ + w₂·z₂) / (w₁ + w₂)

onde w₁ e w₂ são as forças de ativação das regras 1 e 2, z₁ = 2, z₂ = 8.

## Exemplo 2: TSK de ordem um

Considere as regras:

1. Se X é Baixo e Y é Baixo então z = 2·x + 3·y + 1  
2. Se X é Alto e Y é Alto então z = -x + 4·y + 5

**Passos:**
1. Fuzzifique as entradas X e Y para obter os graus de pertinência em "Baixo" e "Alto".
2. Calcule a força de ativação de cada regra (por exemplo, mínimo entre os graus de pertinência dos antecedentes).
3. Calcule o valor do consequente de cada regra usando as funções lineares e os valores de entrada.
4. A saída final é a média ponderada dos valores calculados, ponderada pelas forças de ativação.

**Fórmula da saída:**

z = (w₁·f₁(x, y) + w₂·f₂(x, y)) / (w₁ + w₂)

onde f₁(x, y) = 2·x + 3·y + 1, f₂(x, y) = -x + 4·y + 5, e w₁, w₂ são as forças de ativação das regras.

## Resumo dos Passos para Inferência TSK

1. **Fuzzificação:** Calcule os graus de pertinência das entradas para cada conjunto fuzzy dos antecedentes.
2. **Força de ativação:** Combine os graus de pertinência dos antecedentes de cada regra (usando T-norma, por exemplo, mínimo).
3. **Avaliação do consequente:** Calcule o valor da função consequente de cada regra usando os valores de entrada.
4. **Agregação:** Calcule a saída final como a média ponderada dos consequentes, ponderada pelas forças de ativação.

Esse processo elimina a etapa de defuzzificação tradicional, pois a saída já é um valor escalar.

## Exemplo Real: Controle de Aquecedor com TSK de Ordem Um

Vamos considerar um sistema de controle de temperatura de um aquecedor, onde:

- **Entradas:**  
  - x = Temperatura atual (°C)  
  - y = Variação da temperatura (°C/min)

- **Saída:**  
  - z = Potência do aquecedor (%)

### Conjuntos fuzzy dos antecedentes

- Temperatura (x): Baixa, Alta
- Variação (y): Diminuindo, Aumentando

### Regras TSK de ordem um

1. **Se Temperatura é Baixa e Variação é Diminuindo então z = 80 - 2·x + 0.5·y**
2. **Se Temperatura é Alta e Variação é Aumentando então z = 20 - x - y**

### Suponha os valores de entrada:

- Temperatura atual: x = 18°C
- Variação: y = -1°C/min

### Passos:

1. **Fuzzificação:**  
   - Grau de pertinência de x=18 em "Baixa": 0.8  
   - Grau de pertinência de x=18 em "Alta": 0.2  
   - Grau de pertinência de y=-1 em "Diminuindo": 0.9  
   - Grau de pertinência de y=-1 em "Aumentando": 0.1

2. **Força de ativação das regras:**  
   - Regra 1: w₁ = min(0.8, 0.9) = 0.8  
   - Regra 2: w₂ = min(0.2, 0.1) = 0.1

3. **Avaliação dos consequentes:**  
   - Regra 1: z₁ = 80 - 2·18 + 0.5·(-1) = 80 - 36 - 0.5 = 43.5  
   - Regra 2: z₂ = 20 - 18 - (-1) = 20 - 18 + 1 = 3

4. **Agregação (saída final):**  
   z = (w₁·z₁ + w₂·z₂) / (w₁ + w₂)  
   z = (0.8·43.5 + 0.1·3) / (0.8 + 0.1)  
   z = (34.8 + 0.3) / 0.9 = 35.1 / 0.9 ≈ 39

**Resultado:**  
A potência recomendada para o aquecedor é aproximadamente **39%**.

```python
import numpy as np
import matplotlib.pyplot as plt

# Funções de pertinência para "Baixa" e "Alta" (exemplo simples)
def pertinencia_baixa(x):
    if x <= 15:
        return 1.0
    elif 15 < x < 25:
        return (25 - x) / 10
    else:
        return 0.0

def pertinencia_alta(x):
    if x <= 15:
        return 0.0
    elif 15 < x < 25:
        return (x - 15) / 10
    else:
        return 1.0

# Funções de pertinência para "Diminuindo" e "Aumentando"
def pertinencia_diminuindo(y):
    if y <= -2:
        return 1.0
    elif -2 < y < 0:
        return (0 - y) / 2
    else:
        return 0.0

def pertinencia_aumentando(y):
    if y <= 0:
        return 0.0
    elif 0 < y < 2:
        return y / 2
    else:
        return 1.0

# Funções consequentes das regras
def regra1(x, y):
    return 80 - 2*x + 0.5*y

def regra2(x, y):
    return 20 - x - y

# Geração dos dados para o gráfico
x = 18
y_values = np.linspace(-3, 3, 100)
saida = []

for y in y_values:
    mu_baixa = pertinencia_baixa(x)
    mu_alta = pertinencia_alta(x)
    mu_diminuindo = pertinencia_diminuindo(y)
    mu_aumentando = pertinencia_aumentando(y)
    w1 = min(mu_baixa, mu_diminuindo)
    w2 = min(mu_alta, mu_aumentando)
    z1 = regra1(x, y)
    z2 = regra2(x, y)
    if (w1 + w2) > 0:
        z = (w1*z1 + w2*z2) / (w1 + w2)
    else:
        z = 0
    saida.append(z)

plt.figure(figsize=(8,5))
plt.plot(y_values, saida, label="Potência recomendada (%)")
plt.xlabel("Variação da temperatura (°C/min)")
plt.ylabel("Potência do aquecedor (%)")
plt.title("Saída do sistema TSK para x = 18°C")
plt.grid(True)
plt.legend()
plt.show()
```

---
### Fórmulas matemáticas principais

#### Força de ativação de uma regra (AND com T-norma mínimo):
$$
w_i = \min(\mu_{A_i}(x), \mu_{B_i}(y))
$$

#### Saída do sistema TSK (ordem zero ou um):
$$
z = \frac{\sum_{i=1}^N w_i \cdot f_i(x, y)}{\sum_{i=1}^N w_i}
$$

- Para TSK ordem zero: $f_i(x, y) = c_i$ (constante)
- Para TSK ordem um: $f_i(x, y) = a_i x + b_i y + c_i$ (linear)

#### Saída do sistema Mamdani (após defuzzificação, centroide):
$$
z = \frac{\int z \cdot \mu_{saida}(z) \, dz}{\int \mu_{saida}(z) \, dz}
$$

Onde $\mu_{saida}(z)$ é a função de pertinência agregada da saída.
