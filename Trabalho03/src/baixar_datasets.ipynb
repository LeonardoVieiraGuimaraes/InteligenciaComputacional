# Download dos Datasets Adult e DryBean

Este notebook faz o download automático dos datasets Adult e DryBean e salva os arquivos na pasta `Trabalho03/data`. Execute este notebook antes de rodar os notebooks de classificação.

```python
import os
import urllib.request
import zipfile

data_dir = "data"
os.makedirs(data_dir, exist_ok=True)

# Baixar Adult
def download_adult():
    adult_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"
    adult_test_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.test"
    adult_names_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.names"
    adult_data_path = os.path.join(data_dir, "adult.data")
    adult_test_path = os.path.join(data_dir, "adult.test")
    adult_names_path = os.path.join(data_dir, "adult.names")
    if not os.path.exists(adult_data_path):
        print("Baixando adult.data...")
        urllib.request.urlretrieve(adult_url, adult_data_path)
    else:
        print("adult.data já existe.")
    if not os.path.exists(adult_test_path):
        print("Baixando adult.test...")
        urllib.request.urlretrieve(adult_test_url, adult_test_path)
    else:
        print("adult.test já existe.")
    if not os.path.exists(adult_names_path):
        print("Baixando adult.names...")
        urllib.request.urlretrieve(adult_names_url, adult_names_path)
    else:
        print("adult.names já existe.")

# Baixar DryBean
def download_drybean():
    drybean_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00602/DryBeanDataset.zip"
    drybean_zip_path = os.path.join(data_dir, "DryBeanDataset.zip")
    drybean_csv_path = os.path.join(data_dir, "Dry_Bean_Dataset.csv")
    if not os.path.exists(drybean_csv_path):
        print("Baixando DryBeanDataset.zip...")
        urllib.request.urlretrieve(drybean_url, drybean_zip_path)
        print("Extraindo Dry_Bean_Dataset.csv...")
        with zipfile.ZipFile(drybean_zip_path, 'r') as zip_ref:
            zip_ref.extractall(data_dir)
        print("Removendo arquivo zip...")
        os.remove(drybean_zip_path)
    else:
        print("Dry_Bean_Dataset.csv já existe.")

# Executar downloads
download_adult()
download_drybean()
print("Download dos datasets concluído.")
```
